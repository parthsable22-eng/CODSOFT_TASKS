```python
#checking unique values
for column in data.select_dtypes(include='object').columns:
    print("\n", column)
    print(data[column].dropna().unique()[:30])
```

```python
#clean spaces
for column in data.select_dtypes(include=['object', 'string']).columns:
    data[column] = data[column].str.strip()
```

```python
#standardize product name
data['Product']
data['Product'].str.title()
```

```python
#standardize categories
data['Category']
data['Category'].str.title()
```

```python
#standardize sales channel
data['Sales_Channel']
data['Sales_Channel'].str.title()
```

```python
#standardize order status
data['Order_Status'] = data['Order_Status'].str.strip().str.lower()

data['Order_Status'] = data['Order_Status'].replace({
    'complete': 'Completed',
    'completed': 'Completed',
    'cancelled': 'Cancelled',
    'pending': 'Pending'
})

data['Order_Status'].value_counts()
```

```python
#convert Date
data['Order_Date'] = pd.to_datetime(
    data['Order_Date'],
    errors='coerce'
)
```

```python
print("Invalid/Missing Dates:", data['Order_Date'].isna().sum())
```

```python
#clean Quantity
data['Quantity'] = data['Quantity'].replace({
    'five': '5'
})

data['Quantity'] = pd.to_numeric(
    data['Quantity'],
    errors='coerce'
)
```

```python
#clean unit price
def clean_price(value):
    if pd.isna(value):
        return np.nan
    
    value = str(value).strip().lower().replace(',', '')
    
    if value == 'unknown':
        return np.nan
    
    if value.endswith('k'):
        return float(value[:-1]) * 1000
    
    try:
        return float(value)
    except:
        return np.nan

data['Unit_Price'] = data['Unit_Price'].apply(clean_price)
```

```python
data['Unit_Price'].head(20)
```

```python
#clean Revenue
data['Revenue'] = pd.to_numeric(
    data['Revenue'],
    errors='coerce'
)
```

```python
#clean Customer Age
data['Customer_Age'] = pd.to_numeric(
    data['Customer_Age'],
    errors='coerce'
)
```

```python
#Handling missing values

#For Categorial columns
categorical_columns = [
    'Customer_Name',
    'Category',
    'City',
    'Payment_Method'
]

for column in categorical_columns:
    data[column] = data[column].fillna('Unknown')

#For numerical columns
numerical_columns = [
    'Quantity',
    'Unit_Price',
    'Revenue',
    'Customer_Age'
]

for column in numerical_columns:
    data[column] = data[column].fillna(data[column].median())

#For Date
data['Order_Date'] = data['Order_Date'].fillna(
    data['Order_Date'].mode()[0]
)
```

```python
#Remove Duplicates
data = data.drop_duplicates()

print("Rows after removing duplicates:", len(data))
```

```python
#Check Final Dataset
print("Missing values:")
print(data.isnull().sum())

print("\nDuplicate rows:")
print(data.duplicated().sum())
```

```python
#Verify Datatypes
data.dtypes
```

```python
#Save Cleaned Dataset
data.to_csv(
    "cleaned_sales_data.csv",
    index=False
)
```
