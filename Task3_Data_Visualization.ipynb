```python
#Load Data
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("cleaned_sales_data.csv")
```

```python
#Graph1 - Revenue by Category

category_revenue = data.groupby(
    'Category'
)['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)

plt.show()
```

```python
#Graph2 - Revenue by Product

product_revenue = data.groupby(
    'Product'
)['Revenue'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

product_revenue.plot(kind='bar')

plt.title('Revenue by Product')
plt.xlabel('Product')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

plt.show()
```

```python
#Graph3 - Monthly Revenue

data['Order_Date'] = pd.to_datetime(data['Order_Date'])

data['Month'] = data['Order_Date'].dt.to_period('M')

monthly_revenue = data.groupby('Month')['Revenue'].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker='o'
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
#Graph4 - Sales Channel Distribution

channel_counts = data['Sales_Channel'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    channel_counts.values,
    labels=channel_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Channel')

plt.show()
```

```python
#Graph5 - Revenue Distribution

plt.figure(figsize=(8, 5))

plt.hist(
    data['Revenue'],
    bins=30
)

plt.title('Revenue Distribution')
plt.xlabel('Revenue')
plt.ylabel('Number of Orders')

plt.show()
```

```python
#Graph6 - Quantity vs Revenue

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=data,
    x='Quantity',
    y='Revenue'
)

plt.title('Quantity vs Revenue')
plt.xlabel('Quantity')
plt.ylabel('Revenue')

plt.show()
```

```python
#Graph7 - Revenue By Order Status

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=data,
    x='Order_Status',
    y='Revenue'
)

plt.title('Revenue Distribution by Order Status')
plt.xlabel('Order Status')
plt.ylabel('Revenue')

plt.show()
```

```python
#Graph8 - Correlation Heatmap

plt.figure(figsize=(8, 6))

numeric_cols = [
    'Quantity',
    'Unit_Price',
    'Revenue',
    'Customer_Age'
]

sns.heatmap(
    data[numeric_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

## Visualization Insights

1. The category-level visualization shows which product categories contribute most to overall revenue.

2. Product-level analysis helps identify the highest and lowest revenue-generating products.

3. The monthly revenue line chart highlights changes in sales performance over time.

4. The sales channel pie chart shows the relative contribution of online and store sales.

5. The revenue histogram shows the distribution and spread of order revenue.

6. The scatter plot helps examine the relationship between quantity sold and revenue.

7. The box plot helps identify differences and potential outliers in revenue across order statuses.

8. The correlation heatmap provides an overview of relationships among numerical variables.
